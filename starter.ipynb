### 1. Extract the data from the hosting service

In the cell below, define a function `extract_cifar_data` that extracts python version of the CIFAR-100 dataset. The CIFAR dataaset is open source and generously hosted by the University of Toronto at: https://www.cs.toronto.edu/~kriz/cifar-100-python.tar.gz 

````python
import requests

def extract_cifar_data(url, filename="cifar.tar.gz"):
    """A function for extracting the CIFAR-100 dataset and storing it as a gzipped file
    
    Arguments:
    url      -- the URL where the dataset is hosted
    filename -- the full path where the dataset will be written
    
    """
    
    # Todo: request the data from the data url
    # Hint: use `requests.get` method
    r = requests.get(url)
    r.raise_for_status()
    with open(filename, "wb") as file_context:
        file_context.write(r.content)
    return
````

For a simple gut-check, let's transform one of our images. Each 1024 items in a row is a channel (red, green, then blue). Each 32 items in the channel are a row in the 32x32 image. Using python, we can stack these channels into a 32x32x3 array, and save it as a PNG file:

````python
import numpy as np

# Each 1024 in a row is a channel (red, green, then blue)
row = dataset_train[b'data'][2]
red, green, blue = row[0:1024], row[1024:2048], row[2048:]

# Each 32 items in the channel are a row in the 32x32 image
red = red.reshape(32,32)
green = green.reshape(32,32)
blue = blue.reshape(32,32)

# Combine the channels into a 32x32x3 image!
combined = np.dstack((red,green,blue))
````

Now that you know how to reshape the images, save them as files, and capture their filenames and labels, let's just capture all the bicycles and motorcycles and save them. Scones Unlimited can use a model that tells these apart to route delivery drivers automatically.

In the following cell, identify the label numbers for Bicycles and Motorcycles:

````python
import pandas as pd                                              # tables (used in the next cells)

# Todo: Filter the dataset_train and dataset_meta objects to find the label numbers for Bicycle and Motorcycles
label_names = [name.decode("utf-8")                              # b'bicycle' -> 'bicycle' (bytes -> text)
               for name in dataset_meta[b'fine_label_names']]    # all 100 names; position = label number

for name in ["bicycle", "motorcycle"]:
    label = label_names.index(name)                              # position of the name = its label number
    count = dataset_train[b'fine_labels'].count(label)           # how many training images have this label
    print(f"{name}: label {label}, {count} training images")
````

Good job! We only need objects with label 8 and 48 -- this drastically simplifies our handling of the data! Below we construct a dataframe for you, and you can safely drop the rows that don't contain observations about bicycles and motorcycles. Fill in the missing lines below to drop all other rows:

````python
# Construct the dataframe
df_train = pd.DataFrame({
    "filenames": dataset_train[b'filenames'],          # file name of each image (bytes)
    "labels": dataset_train[b'fine_labels'],           # label number of each image (0-99)
    "row": range(len(dataset_train[b'filenames']))     # position of the image in b'data' (0..49999)
})

# Drop all rows from df_train where label is not 8 or 48
df_train = df_train[df_train["labels"].isin([8, 48])]  # keep only bicycles (8) and motorcycles (48)

# Decode df_train.filenames so they are regular strings
df_train["filenames"] = df_train["filenames"].apply(
    lambda x: x.decode("utf-8")                        # b'bicycle_s_000017.png' -> 'bicycle_s_000017.png'
)

df_test = pd.DataFrame({
    "filenames": dataset_test[b'filenames'],
    "labels": dataset_test[b'fine_labels'],
    "row": range(len(dataset_test[b'filenames']))      # position in dataset_test[b'data'] (0..9999)
})

# Drop all rows from df_test where label is not 8 or 48
df_test = df_test[df_test["labels"].isin([8, 48])]     # same filter for the test set

# Decode df_test.filenames so they are regular strings
df_test["filenames"] = df_test["filenames"].apply(
    lambda x: x.decode("utf-8")
)
````

Compose these together into a function that saves all the images into the `./test` and `./train` directories. Use the comments in the body of the `save_images` function below to guide your construction of the function:


````python
def save_images(dataset, path, row):
    """Save one image as a PNG file.
    dataset -- dataset_train or dataset_test (where the pixels are)
    path    -- target folder, e.g. "./train/"
    row     -- one row of df_train / df_test (filenames, labels, row)
    """
    # Grab the image data in row-major form
    img = dataset[b'data'][row["row"]]                  # 3072 numbers of this image, found via the "row" column

    # Consolidated stacking/reshaping from earlier
    target = np.dstack((
        img[0:1024].reshape(32, 32),                    # red layer
        img[1024:2048].reshape(32, 32),                 # green layer
        img[2048:].reshape(32, 32)))                    # blue layer -> shape (32, 32, 3)

    # Save the image
    plt.imsave(path + row["filenames"], target)         # e.g. ./train/bicycle_s_000017.png

    # Return any signal data you want for debugging
    return path + row["filenames"]                      # the saved path (handy for checking)

## TODO: save ALL images using the save_images function
for _, row in df_train.iterrows():                      # go through all 1,000 training rows
    save_images(dataset_train, "./train/", row)         # pixels from dataset_train -> ./train/

for _, row in df_test.iterrows():                       # go through all 200 test rows
    save_images(dataset_test, "./test/", row)           # pixels from dataset_test -> ./test/
````

## Model Training

For Image Classification, Sagemaker [also expects metadata](https://docs.aws.amazon.com/sagemaker/latest/dg/image-classification.html) e.g. in the form of TSV files with labels and filepaths. We can generate these using our Pandas DataFrames from earlier:

````python
def to_metadata_file(df, prefix):
    df["s3_path"] = df["filenames"]
    df["labels"] = df["labels"].apply(lambda x: 0 if x==8 else 1)
    return df[["row", "labels", "s3_path"]].to_csv(
        f"{prefix}.lst", sep="\t", index=False, header=False
    )
    
to_metadata_file(df_train.copy(), "train")
to_metadata_file(df_test.copy(), "test")
````

Using the `bucket` and `region` info we can get the latest prebuilt container to run our training job, and define an output location on our s3 bucket for the model. Use the `image_uris` function from the SageMaker SDK to retrieve the latest `image-classification` image below:

````python
# Use the image_uris function to retrieve the latest 'image-classification' image 
from sagemaker import image_uris

algo_image = image_uris.retrieve("image-classification", region)   # AWS's built-in image classifier (ECR, us-east-1)
s3_output_location = f"s3://{bucket}/models/image_model"          # where model.tar.gz will be saved
print(algo_image)                                                 # e.g. 811284229777.dkr.ecr.us-east-1.amazonaws.com/image-classification:1
````

We're ready to create an estimator! Create an estimator `img_classifier_model` that uses one instance of `ml.p3.2xlarge`. Ensure that y ou use the output location we defined above - we'll be referring to that later!

````python
img_classifier_model = sagemaker.estimator.Estimator(
    image_uri=algo_image,                   # built-in image classification container
    role=role,                              # role for S3 access
    instance_count=1,                       # one training machine
    instance_type="ml.g4dn.xlarge",         # FIX: current GPU (NVIDIA T4) instead of the retired ml.p3.2xlarge
    output_path=s3_output_location,         # model.tar.gz goes here
    max_run=1800,                           # safety cap: 30 min
    sagemaker_session=session)
````

We can also set a few key hyperparameters and define the inputs for our model:

````python
img_classifier_model.set_hyperparameters(
    image_shape="3,32,32",                  # channels, height, width -> RGB 32x32 images
    num_classes=2,                          # bicycle (0) and motorcycle (1)
    num_training_samples=len(df_train))     # 1000 -> the algorithm needs this to plan its epochs
````

The `image-classification` image uses four input channels with very specific input parameters. For convenience, we've provided them below:

````python
from sagemaker.debugger import Rule, rule_configs
from sagemaker.session import TrainingInput
model_inputs = {
        "train": sagemaker.inputs.TrainingInput(
            s3_data=f"s3://{bucket}/train/",
            content_type="application/x-image"
        ),
        "validation": sagemaker.inputs.TrainingInput(
            s3_data=f"s3://{bucket}/test/",
            content_type="application/x-image"
        ),
        "train_lst": sagemaker.inputs.TrainingInput(
            s3_data=f"s3://{bucket}/train.lst",
            content_type="application/x-image"
        ),
        "validation_lst": sagemaker.inputs.TrainingInput(
            s3_data=f"s3://{bucket}/test.lst",
            content_type="application/x-image"
        )
}
````

Great, now we can train the model using the model_inputs. In the cell below, call the `fit` method on our model,:

````python
img_classifier_model.fit(model_inputs)        # starts the training job and waits (~5-10 min, prints logs)
````

### Testing and Evaluation

Do several step function invokations using data from the `./test` folder. This process should give you confidence that the workflow both *succeeds* AND *fails* as expected. In addition, SageMaker Model Monitor will generate recordings of your data and inferences which we can visualize.

Here's a function that can help you generate test inputs for your invokations:

````python
import random
import boto3
import json

def generate_test_case():
    # Setup s3 in boto3
    s3 = boto3.resource('s3')

    # Randomly pick from the test folder in our bucket
    objects = s3.Bucket(bucket).objects.filter(Prefix="test/")   # FIX: filter needs Prefix=..., not a plain string

    # Grab any random object key from that folder!
    obj = random.choice([x.key for x in objects])

    return json.dumps({
        "image_data": "",
        "s3_bucket": bucket,
        "s3_key": obj
    })

generate_test_case()          # e.g. '{"image_data": "", "s3_bucket": "...", "s3_key": "test/motorcycle_s_000141.png"}'
````

````python
import time

sfn = boto3.client("stepfunctions")
STATE_MACHINE_ARN = f"arn:aws:states:{region}:{bucket.split('-')[-1]}:stateMachine:scones-image-classification"   # account id taken from the bucket name

for i in range(20):                                        # 20 random test images
    sfn.start_execution(stateMachineArn=STATE_MACHINE_ARN, input=generate_test_case())
    time.sleep(1)                                          # small gap so the inferences spread over time
print("started 20 executions")
````

````python
from collections import Counter

runs = sfn.list_executions(stateMachineArn=STATE_MACHINE_ARN, maxResults=50)["executions"]
print(Counter(r["status"] for r in runs))                  # e.g. Counter({'SUCCEEDED': 15, 'FAILED': 6})
````

The data are in JSONLines format, where multiple valid JSON objects are stacked on top of eachother in a single `jsonl` file. We'll import an open-source library, `jsonlines` that was purpose built for parsing this format.

Now we can extract the data from each of the source files:

````python
import glob

# List the file names we downloaded (they sit in date sub-folders, so search recursively)
file_handles = glob.glob("./captured_data/**/*.jsonl", recursive=True)      # FIX 1: os.listdir misses sub-folders

# Dump all the data into an array
json_data = []
for jsonl in file_handles:
    with jsonlines.open(jsonl) as f:
        for obj in f:                         # FIX 2: a file can hold MANY requests -> read every line, not just the first
            json_data.append(obj)

print(len(file_handles), "files,", len(json_data), "captured inferences")
````

The data should now be a list of dictionaries, with significant nesting. We'll give you an example of some code that grabs data out of the objects and visualizes it:

````python
# Define how we'll get our data
def simple_getter(obj):
    inferences = obj["captureData"]["endpointOutput"]["data"]
    timestamp = obj["eventMetadata"]["inferenceTime"]
    return json.loads(inferences), timestamp

simple_getter(json_data[0])
````

Finally, here's an example of a visualization you can build with this data. In this last part, you will take some time and build your own - the captured data has the input images, the resulting inferences, and the timestamps.

````python
import matplotlib.pyplot as plt
import pandas as pd

THRESHOLD = 0.93                                          # FIX 1: same threshold as Lambda 3 (Udacity's chart used .94)

# Populate the data for the x and y axis
x, y = [], []
for obj in json_data:
    inference, timestamp = simple_getter(obj)
    y.append(max(inference))                              # confidence = probability of the predicted class
    x.append(pd.to_datetime(timestamp))                   # text -> real time, so the x-axis is ordered correctly

plt.style.use("Solarize_Light2")
plt.figure(figsize=(9, 4))
plt.scatter(x, y, c=['r' if k < THRESHOLD else 'b' for k in y])   # red = would fail the workflow, blue = passes
plt.axhline(y=THRESHOLD, color='g', linestyle='--')                # the threshold line
plt.ylim(bottom=0.5)                                      # FIX 2: Udacity's .88 hid every low-confidence point
plt.ylabel("Confidence")
plt.suptitle("Observed Recent Inferences", size=14)
plt.title("Pictured with confidence threshold for production use", size=10)
plt.gcf().autofmt_xdate()
plt.show()
````

````python
import base64, hashlib, glob, os

# 1) true labels from test.lst (columns: index, label 0/1, file name)
lst = pd.read_csv("test.lst", sep="\t", header=None, names=["row", "label", "fname"])
true_label = dict(zip(lst["fname"], lst["label"].map({0: "bicycle", 1: "motorcycle"})))   # file name -> true class

# 2) fingerprint every local test image -> file name
fingerprint = {}
for path in glob.glob("./test/*.png"):
    with open(path, "rb") as f:
        fingerprint[hashlib.md5(f.read()).hexdigest()] = os.path.basename(path)

# 3) one row per captured inference
rows = []
for obj in json_data:
    inference, timestamp = simple_getter(obj)
    img_bytes = base64.b64decode(obj["captureData"]["endpointInput"]["data"])      # captured input image
    fname = fingerprint.get(hashlib.md5(img_bytes).hexdigest())                     # matching test file (or None)
    rows.append({
        "time": pd.to_datetime(timestamp),
        "file": fname,
        "confidence": max(inference),
        "predicted": "bicycle" if inference[0] > inference[1] else "motorcycle",    # index 0 = bicycle
        "true": true_label.get(fname),                                              # FIX: real label, not the file-name prefix
    })

monitor = pd.DataFrame(rows)
monitor["passed"] = monitor["confidence"] >= THRESHOLD
monitor["correct"] = monitor["predicted"] == monitor["true"]
monitor     # expect only bicycle / motorcycle     # only 'bicycle' / 'motorcycle' (plus None for the Lambda test, if any)
````

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Left: how confident is the model, per predicted class?
for cls, color in [("bicycle", "tab:blue"), ("motorcycle", "tab:orange")]:
    axes[0].hist(monitor.loc[monitor["predicted"] == cls, "confidence"],
                 bins=10, range=(0.5, 1.0), alpha=0.6, label=cls, color=color)
axes[0].axvline(THRESHOLD, color="g", linestyle="--", label=f"threshold {THRESHOLD}")
axes[0].set_xlabel("Confidence"); axes[0].set_ylabel("Number of inferences")
axes[0].set_title("Confidence distribution by predicted class"); axes[0].legend()

# Right: are the predictions that PASS the threshold actually correct?
known = monitor.dropna(subset=["true"])                                    # only rows with a recovered true label
summary = known.groupby("passed")["correct"].agg(["mean", "count"])
labels = [f"{'passed' if p else 'rejected'}\n(n={n})" for p, n in zip(summary.index, summary["count"])]
axes[1].bar(labels, summary["mean"], color=["tab:red" if not p else "tab:green" for p in summary.index])
axes[1].set_ylim(0, 1); axes[1].set_ylabel("Accuracy")
axes[1].set_title("Accuracy of rejected vs. passed inferences")

plt.suptitle("Scones Unlimited – endpoint monitoring from captured data", size=14)
plt.tight_layout(); plt.show()

print(f"Pass rate: {monitor['passed'].mean():.0%} | Live accuracy (all): {known['correct'].mean():.0%}")
````
